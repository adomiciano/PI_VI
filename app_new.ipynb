```python
import os
import cv2
from ultralytics import YOLO

# Configuração dos caminhos
print("Carregando modelo YOLOv8...")
# Use seu modelo customizado:
MODEL_PATH = "C:/Users/domic/downloads/best.pt"
model = YOLO(MODEL_PATH)

# Ou para testar:
# model = YOLO('yolov8n.pt')

print("Modelo carregado com sucesso!")

# Inicializa a webcam (0 = webcam padrão)
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("ERRO: Não foi possível abrir a webcam!")
    exit()

print("Webcam iniciada. Pressione 'q' para sair.")

while True:
    ret, frame = cap.read()
    
    if not ret:
        print("Erro ao capturar frame da webcam")
        break
    
    # Realiza a detecção
    results = model(frame)[0]
    
    # Verifica se detectou pessoa e capacete
    has_person = False
    has_hardhat = False
    detected_objects = []
    
    for result in results.boxes.data.tolist():
        x1, y1, x2, y2, score, class_id = result
        class_name = results.names[int(class_id)]
        
        if score > 0.5:
            detected_objects.append(f"{class_name}: {score:.2f}")
        
        if class_name == 'person':
            has_person = True
        if class_name == 'hardhat':
            has_hardhat = True
        
        if score > 0.6:
            cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)
            cv2.putText(frame, f"{class_name.upper()} {score:.2f}", 
                       (int(x1), int(y1 - 10)),
                       cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
    
    # Determina o status
    if has_person and not has_hardhat:
        status = "ACESSO NEGADO: Sem capacete"
        color = (0, 0, 255)
    elif has_person and has_hardhat:
        status = "ACESSO LIBERADO: Com capacete"
        color = (0, 255, 0)
    elif not has_person and has_hardhat:
        status = "ACESSO LIBERADO: Somente capacete"
        color = (0, 255, 0)
    else:
        if detected_objects:
            status = f"Detectado: {', '.join(detected_objects[:3])}"
        else:
            status = "Aguardando detecção..."
        color = (255, 255, 0)
    
    # Exibe o status
    cv2.rectangle(frame, (10, 10), (620, 60), (0, 0, 0), -1)
    cv2.putText(frame, status, (20, 45), 
                cv2.FONT_HERSHEY_SIMPLEX, 1, color, 2)
    
    cv2.imshow('Detecção de Capacete - Pressione Q para sair', frame)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
print("Webcam fechada.")
```
